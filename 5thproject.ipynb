### CONVERTING VOLUME COLUMN INTO NUMBERS

```python
import numpy as np
import pandas as pd

def convert_volume(val):
    # Check for empty values or the literal dash placeholder
    if pd.isna(val) or val == '-':
        return np.nan

    val = str(val).strip().upper()
    if 'K' in val:
        return float(val.replace('K', '')) * 1_000
    elif 'M' in val:
        return float(val.replace('M', '')) * 1_000_000
    elif 'B' in val:
        return float(val.replace('B', '')) * 1_000_000_000

    return float(val)

# Apply the transformation to the 'Vol.' column
df['Vol.'] = df['Vol.'].apply(convert_volume)

# Now, fill the missing values with the median of the converted numeric 'Vol.' column
df['Vol.'] = df['Vol.'].fillna(df['Vol.'].median())
df
```

### Train-test split using time-based split

```python
import pandas as pd

# Ensure df is loaded in case previous cells were not run or kernel was reset
df = pd.read_csv("Bitcoin_Historical_Data_Sorted.csv")

# Define your split ratio (e.g., 80% for training)
train_ratio = 0.80
split_idx = int(len(df) * train_ratio)

# 3. Slice the data chronologically without shuffling
train_df = df.iloc[:split_idx]
# The remaining 20% goes to testing
test_df = df.iloc[split_idx:]

# Quick verification check
print(f"Total records: {len(df)}")
print(f"Train set:     {len(train_df)} rows (from {train_df.index.min()} to {train_df.index.max()})")
print(f"Test set:      {len(test_df)} rows (from {test_df.index.min()} to {test_df.index.max()})")
```

### Selecting features and target

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 1. Select features and target (e.g., using 'Price' to predict the next day's 'Price')
# Adjust column names based on your actual CSV
features = ['Price']  # Add other features if you have them (e.g., 'Volume', 'Open')
target = 'Price'

# Fix: Ensure the 'Price' column in train_df and test_df is numeric before scaling
# It was reloaded as a CSV in the previous cell, losing the float conversion.
train_df.loc[:, 'Price'] = train_df['Price'].astype(str).str.replace(',', '', regex=False).astype(float)
test_df.loc[:, 'Price'] = test_df['Price'].astype(str).str.replace(',', '', regex=False).astype(float)

# 2. Scale the data (Fit ONLY on train set to prevent data leakage)
scaler = MinMaxScaler(feature_range=(0, 1))
# Fix: Fit scaler on the values array to avoid the feature name warning later
train_scaled = scaler.fit_transform(train_df[features].values)
test_scaled = scaler.transform(test_df[features].values)

# 3. Multi-Horizon Sliding Window Function
def create_multi_horizon_sequences(data, lookback):
    X, y = [], []

    # The + 7 accounts for the 7-day ahead prediction, ensuring enough data points exist
    for i in range(len(data) - lookback - 7):
        X.append(data[i : i + lookback])

        # Collect targets for 1-day, 3-day, and 7-day ahead
        y_1day = data[i + lookback, 0]
        y_3day = data[i + lookback + 2, 0]
        y_7day = data[i + lookback + 6, 0]

        y.append([y_1day, y_3day, y_7day])

    return np.array(X), np.array(y)

# Define lookback window (e.g., use past 60 days)
LOOKBACK = 60

X_train, y_train = create_multi_horizon_sequences(train_scaled, LOOKBACK)
X_test, y_test = create_multi_horizon_sequences(test_scaled, LOOKBACK)

print(f"X_train shape: {X_train.shape}") # Expected: (samples, 60, 1)
print(f"y_train shape: {y_train.shape}") # Expected: (samples, 3)

print(f"X_test shape: {X_test.shape}")   # Expected: (samples, 60, 1)
print(f"y_test shape: {y_test.shape}")    # Expected: (samples, 3)
```

### 1D CNN MODEL

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Flatten, Dense, LeakyReLU, Dropout, Input
from tensorflow.keras.optimizers import Adam

# Define the forecasting horizons
HORIZONS = [1, 3, 7] # For 1-day, 3-day, and 7-day ahead predictions

# --- Model Architecture ---
model_cnn = Sequential([
    Input(shape=(LOOKBACK, X_train.shape[2])), # Explicit Input Layer
    Conv1D(filters=64, kernel_size=3),
    LeakyReLU(negative_slope=0.1),
    Dropout(0.2), # Prevents overfitting

    Conv1D(filters=32, kernel_size=3),
    LeakyReLU(negative_slope=0.1),
    Dropout(0.2),

    Flatten(), # Flattening avoids the loss of temporal data from Global Pooling

    Dense(32),
    LeakyReLU(negative_slope=0.1),

    # Dense output layer must match the number of horizons (3: 1-day, 3-day, 7-day)
    Dense(len(HORIZONS))
])

# --- Compile Model ---
# Using a slightly lower learning rate (0.0005) to keep the gradients stable
model_cnn.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='mse',
    metrics=['mae']
)

model_cnn.summary()

# --- Train Model ---
print("\n--- Starting Model Training ---")
history = model_cnn.fit(
    X_train, y_train,
    epochs=40,
    batch_size=32,
    validation_split=0.1,  # Uses 10% of training data to check for overfitting
    verbose=1
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# 1. Predict on test set using the trained model_cnn
# Ensure model_cnn, X_test, y_test, and scaler are defined from previous cells
predictions_scaled = model_cnn.predict(X_test)

# Safe Denormalization Function for Multi-Horizon Targets
def inverse_transform_multi(scaler, scaled_data):
    # Ensure data is numpy array
    scaled_data = np.asarray(scaled_data)

    # Handle both 1D (single horizon) and 2D (multi-horizon) cases for y_test
    if scaled_data.ndim == 1:
        orig_shape = scaled_data.shape
        denorm_flat = scaler.inverse_transform(scaled_data.reshape(-1, 1))
        return denorm_flat.reshape(orig_shape)
    else:
        orig_shape = scaled_data.shape
        # Flatten to a single column, inverse transform, and reshape back
        flat_data = scaled_data.reshape(-1, 1)
        denorm_flat = scaler.inverse_transform(flat_data)
        return denorm_flat.reshape(orig_shape)

predictions_actual = inverse_transform_multi(scaler, predictions_scaled)
y_test_actual = inverse_transform_multi(scaler, y_test)

horizons = ['1-Day Ahead', '3-Day Ahead', '7-Day Ahead']

print("--- Model Evaluation Metrics for CNN ---")

# 3. Loop through columns to evaluate each horizon individually
for i, horizon_name in enumerate(horizons):
    actuals = y_test_actual[:, i]
    preds = predictions_actual[:, i]

    mae = mean_absolute_error(actuals, preds)
    rmse = np.sqrt(mean_squared_error(actuals, preds))

    # Small epsilon added to denominator to prevent division by zero
    mape = np.mean(np.abs((actuals - preds) / (actuals + np.finfo(float).eps))) * 100

    print(f"\n📈 Metrics for {horizon_name}:")
    print(f"  MAE:  ${mae:.2f}")
    print(f"  RMSE: ${rmse:.2f}")
    print(f"  MAPE: {mape:.2f}%")
```

# RNN MODEL

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Dropout

# Define the SimpleRNN Architecture
model_rnn = Sequential([
    # First SimpleRNN layer (return_sequences=True to pass sequences to the next layer)
    SimpleRNN(units=64, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.2),

    # Second SimpleRNN layer (return_sequences=False returns only the final state)
    SimpleRNN(units=32, return_sequences=False),
    Dropout(0.2),

    # Fully connected hidden layer
    Dense(32, activation='relu'),

    # Output layer: 3 units for [1-day, 3-day, 7-day] forecasts
    Dense(3)
])

# Compile the SimpleRNN model
model_rnn.compile(optimizer='adam', loss='mse', metrics=['mae'])
model_rnn.summary()

# Train the SimpleRNN model
history_rnn = model_rnn.fit(
    X_train, y_train,
    epochs=30,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Predict on test data using SimpleRNN
predictions_scaled_rnn = model_rnn.predict(X_test)

# 2. Denormalize back to actual Bitcoin prices (USD)
predictions_actual_rnn = scaler.inverse_transform(predictions_scaled_rnn)
y_test_actual = scaler.inverse_transform(y_test)

horizons = ['1-Day Ahead', '3-Day Ahead', '7-Day Ahead']

print("--- Multi-Horizon SimpleRNN Evaluation Metrics ---")

# 3. Calculate metrics loop
for i, horizon_name in enumerate(horizons):
    actuals = y_test_actual[:, i]
    preds = predictions_actual_rnn[:, i]

    mae = mean_absolute_error(actuals, preds)
    rmse = np.sqrt(mean_squared_error(actuals, preds))
    mape = np.mean(np.abs((actuals - preds) / actuals)) * 100

    print(f"\n📊 Metrics for {horizon_name}:")
    print(f"  MAE:  ${mae:.2f}")
    print(f"  RMSE: ${rmse:.2f}")
    print(f"  MAPE: {mape:.2f}%")
```

## LSTM MODEL

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# Define the LSTM Architecture
model_lstm = Sequential([
    # First LSTM layer (return_sequences=True to pass sequences to the next layer)
    LSTM(units=64, return_sequences=True, input_shape=(X_train.shape[1], X_train.shape[2])),
    Dropout(0.2),

    # Second LSTM layer (return_sequences=False returns only the final hidden state)
    LSTM(units=32, return_sequences=False),
    Dropout(0.2),

    # Fully connected hidden layer
    Dense(32, activation='relu'),

    # Output layer: 3 units for [1-day, 3-day, 7-day] forecasts
    Dense(3)
])

# Compile the LSTM model
model_lstm.compile(optimizer='adam', loss='mse', metrics=['mae'])
model_lstm.summary()

# Train the LSTM model
history_lstm = model_lstm.fit(
    X_train, y_train,
    epochs=30,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Predict on test data using LSTM
predictions_scaled_lstm = model_lstm.predict(X_test)

# 2. Denormalize back to actual Bitcoin prices (USD)
predictions_actual_lstm = scaler.inverse_transform(predictions_scaled_lstm)
y_test_actual = scaler.inverse_transform(y_test)

horizons = ['1-Day Ahead', '3-Day Ahead', '7-Day Ahead']

print("--- Multi-Horizon LSTM Evaluation Metrics ---")

# 3. Calculate metrics loop
for i, horizon_name in enumerate(horizons):
    actuals = y_test_actual[:, i]
    preds = predictions_actual_lstm[:, i]

    mae = mean_absolute_error(actuals, preds)
    rmse = np.sqrt(mean_squared_error(actuals, preds))
    mape = np.mean(np.abs((actuals - preds) / actuals)) * 100

    print(f"\n📊 Metrics for {horizon_name}:")
    print(f"  MAE:  ${mae:.2f}")
    print(f"  RMSE: ${rmse:.2f}")
    print(f"  MAPE: {mape:.2f}%")
```

### TRANSFORMER MODEL

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, LayerNormalization, MultiHeadAttention, GlobalAveragePooling1D

# 1. Inputs
inputs = Input(shape=(X_train.shape[1], X_train.shape[2]))

# 2. Transformer Encoder Block
# Multi-Head Self-Attention Layer
attention_output = MultiHeadAttention(num_heads=4, key_dim=64)(inputs, inputs)
attention_output = Dropout(0.2)(attention_output)
x1 = LayerNormalization(epsilon=1e-6)(inputs + attention_output)

# Feed-Forward Network Block
ffn = Sequential([
    Dense(64, activation='relu'),
    Dense(X_train.shape[2])
])(x1)
ffn = Dropout(0.2)(ffn)
x2 = LayerNormalization(epsilon=1e-6)(x1 + ffn)

# 3. Pooling and Prediction Head
x = GlobalAveragePooling1D()(x2)
x = Dense(32, activation='relu')(x)
outputs = Dense(3)(x)  # 3 units for [1-day, 3-day, 7-day] forecasts

# Define, Compile, and Summarize Model
model_transformer = Model(inputs=inputs, outputs=outputs)
model_transformer.compile(optimizer='adam', loss='mse', metrics=['mae'])
model_transformer.summary()

# Train the Transformer Model
history_transformer = model_transformer.fit(
    X_train, y_train,
    epochs=30,
    batch_size=32,
    validation_data=(X_test, y_test),
    verbose=1
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# 1. Generate predictions using the trained Transformer model
predictions_scaled_trans = model_transformer.predict(X_test)

# 2. Inverse transform back to the original Bitcoin price scale (USD)
predictions_actual_trans = scaler.inverse_transform(predictions_scaled_trans)
y_test_actual = scaler.inverse_transform(y_test)

horizons = ['1-Day Ahead', '3-Day Ahead', '7-Day Ahead']

print("--- Multi-Horizon Transformer Evaluation Metrics ---")

# 3. Calculate and print metrics for each horizon
for i, horizon_name in enumerate(horizons):
    actuals = y_test_actual[:, i]
    preds = predictions_actual_trans[:, i]

    mae = mean_absolute_error(actuals, preds)
    rmse = np.sqrt(mean_squared_error(actuals, preds))
    mape = np.mean(np.abs((actuals - preds) / actuals)) * 100

    print(f"\n🤖 Metrics for {horizon_name}:")
    print(f"  MAE:  ${mae:.2f}")
    print(f"  RMSE: ${rmse:.2f}")
    print(f"  MAPE: {mape:.2f}%")
```
